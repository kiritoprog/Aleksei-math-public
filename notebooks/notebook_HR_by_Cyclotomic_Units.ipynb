# The Herbrand–Ribet theorem by cyclotomic untis

Let $\mathbb{Q}(\zeta_p)$ be the cyclotomic field for a prime $p$, and let us denote by $\mathrm{Cl}$ its class group.

Denote by $B_n$ the $n$-th Bernoulli number, defined by the generating function
$$
\sum_{n=0}^{\infty}\frac{B_n}{n!}t^n=\frac{t}{e^t-1}.
$$

Essentially, Bernoulli numbers are the values of the Riemann zeta function at negative integers, since
$$
\zeta(1-n)=-\frac{B_n}{n}.
$$

**Theorem (Kummer)**. $p \mid \# \mathrm{Cl}$ if and only if $p \mid B_i$ for some $i = 0, 2, \ldots, p - 3$.

Such primes are called irregular primes. If we denote by $A$ the $p$-component of $\mathrm{Cl}$, then $p$ is irregular if and only if $A \neq 0$.

But what information do invidual Bernoulli numbers give? There is some natural decomposition $A = \bigoplus\limits_{i \in \mathbb{Z}/(p-1)\mathbb{Z}} A^{(i)}$ which will be explained later, so that the following theorem holds.

**Theorem (Herbrand-Ribet)**. $A^{(i)} \neq 0$ if and only if $p \mid B_{p-i}$ for odd $i \leqslant p$.

The converse direction which gives non-trivial class from divisibility of Bernoulli number is the hardest and the most essential part.

The first irregular prime is $37$.

The script below computes the order of the $p$-part of the class group of $K=\mathbb{Q}(\zeta_p)$.

```gp
p=37;
K = bnfinit(polcyclo(p)); /* p is small, can construct bnf */
```

We can notice a pattern between such primes $p$ and divisibility by $p$ of the Bernoulli numbers $B_i$.

```gp
forstep(ii=2,p-1,2,print(p,"-valuation of ", ii, "-th Bernoulli number: ", valuation(bernfrac(ii),p)))
```

For $p=37$, the Bernoulli number with nontrivial $p$-part has index $32$.

Odd Bernoulli numbers vanish, except for $B_1=-\frac{1}{2}$, so they can't have any meaningful information. The fact that $v_p(B_{p-1})=-1$ is a consequence of the von Staudt--Clausen theorem.

**Theorem(von Staudt-Clausen)**. $B_n + \sum\limits_{p - 1\mid n} \frac{1}{p} \in \mathbb{Z}$ for all even $n$.

How can we provide non-trivial class $[\mathfrak{a}]$ in $A$? Recall basic class field theory.

There is unique unramified abelian extension $H/K$ such that Artin symbol induces isomorphism $Cl(K) \to \mathrm{Gal}(H/K)$. It can be characterized as maximal abelian everywhere unramified extension of $K$.

Thus by Galois theory $\mathbb{Z}/p\mathbb{Z}$ **quotients** of $C$ are just $\mathbb{Z}/p\mathbb{Z}$-extensions of $K$ which are everywhere unramified.

But obviously $\zeta_p \in K$, so by Kummer theory such extensions are always of the form $K(\alpha^{1/p})$. Suppose $\alpha$ is a unit of $K$. Then the only ramification can occur above $1 - \zeta_p$, the only place in $K$ above $p$. This leads to the following

**Observation.** Non-trivial classes can be encoded by units of $K$.

Suppose we are given a unit generating unramified extension as above. How do we extract class from it?
If Artin symbol of $\mathfrak{a}$ is non-zero then the image of the class of $\mathfrak{a}$ is non-trivial in $\mathbb{Z}/p\mathbb{Z}$-quotient corresponding to the extension. By Chebotarev we may assume that the residue field of $\mathfrak{a}$ is $\mathbb{F}_\ell$ for some rational prime $\ell$.

**Proposition**. Suppose $\alpha$ is a unit in $K$, $\mathfrak{a}$ is a prime in $K$ with residue field $\mathbb{F}_\ell$ for $\ell \neq p$. Then the Frobenius of $\mathfrak{a}$ in $K(\alpha^{1/p})$ is non-trivial if and only if reduction $\overline{\alpha} \in \mathbb{F}_\ell$ is not $p$-th root.

```gp
/* Given prime l = 1 (mod p) (i.e completely split in K),
   and unit u in K checks if the Frobenius of some (equivalently every) prime in K
   over l in extension K(u^{1/p})/K is non-trivial */
check(l, u) = {
    g = znprimroot(l);
    zpmodl = g^((l-1)/p); /* choosing pth root mod l amounts to choosing prime above l */
    umodl = subst(lift(u), x, zpmodl);

    return (umodl^((l-1)/p) != 1);
}
```

The only thing we are left to check is that $K(u^{1/p})$ is unramified at $1 - \zeta_p$. How can we do that?

If $u$ is $1-\zeta_p$-adically $p$-th root, then the local extension is obviously unramified.

So we need to check that locally at $p$ it is possible to extract the $p$-th root of $u$. It can be verified for sufficiently close approximations by Hensel's lemma.

**Hensel's Lemma.** Suppose $K$ is a field complete under non-archimedean valuation $\lvert \cdot \rvert$, $f \in \mathcal{O}_K[x]$ is a monic polynomial. If $\lvert f^\prime(x) \rvert^2 < \lvert f(x) \rvert$ for some $x \in \mathcal{O}_K$ then $f$ has a root in $\mathcal{O}_K$.

Applying this lemma to $f = x^p - u$ in completion of $K$ by $1 - \zeta_p$, since $v(p) = p - 1$ we see that it is enough to check that $f$ has a root modulo $(1-\zeta)^{2p-1}$.

```gp
check_plocalpower(u,M) = {
    g = znprimroot(p^M);
    zpmodpM = g^((p-1)*p^(M-1)/p);
    umodpM = subst(lift(u),x,zpmodpM);
    
    return(umodpM^((p-1)*p^(M-1)/p)==1)
}

print(check_plocalpower(u,2*p-1))
```

On the other hand, divisibility of certain Bernoulli numbers provides a root of corresponding $p$-adic $L$-function.

Let us define $p$-adic $L$-function associated with a power of Teichmuller character.

For positive integers $n \equiv i \mod (p-1)$ set:
$$
L_p(1-n,\omega^i) = -(1-p^{n-1})\frac{B_n}{n}.
$$
By the Kummer congruences, and since such integers are dense in $\mathbb{Z}_p$, this formula defines a continuous function.

**Theorem(Kummer)**. For $n \equiv m \mod (p-1)p^N$ there is a congruence
$$
(1-p^{n-1})\frac{B_n}{n} \equiv (1-p^{m-1})\frac{B_m}{m} \mod p^N.
$$

In fact, $L_p(s,\omega^i)$ is a $p$-adic analytic function.

This definition shows that searching for zeros of $L_p(s,\omega^i)$ is directly related to finding Bernoulli numbers whose divisibility by $p$ increases as their indices grow within the congruence class of $i$ modulo $p-1$.

The following script searches for such Bernoulli numbers with indices in a fixed residue class modulo $p-1$ over a small range.

```gp
bernvec(300);

bernoulli_root_appr(p,ip,N) = {
    indices = List();
    for(nn=1,10,forstep(kk=2,2*N,2,\
    if((kk%(p-1)==ip)&(numerator(bernfrac(kk))%p^nn==0),\
    listput(indices,kk);break())));

    return(indices);
}

indices = bernoulli_root_appr(37,32,300)
```

Let $b_{p,i}$ be the limit of a convergent subsequence from the indices founded above. Then the desired zero $s_{p,i}$ of $L_p(s,\omega^i)$ is given by $1-b_{p,i}$.

**Propopsition**. For integer $n \ge 1$
$$
L_p(1-n,\omega^i) = -(1-\omega^{i-n}(p)p^{n-1})\frac{B_{n,\omega^{i-n}}}{n},
$$
where $B_{n,\chi}$ is the $n$-th generalized Bernoulli number associated with character $\chi$ of conductor $f$, defined by $\sum\limits_{n=0}^\infty \frac{B_{n,\chi}}{n!} t^n = \sum\limits_{a=1}^f\frac{\chi(a)te^{at}}{e^{ft}-1}$.

Recall that a non-trivial power of the Teichmüller character vanishes at $p$. Hence, if $n \not\equiv i \mod (p-1)$, the Euler factor $(1-\omega^{i-n}(p)p^{n-1})$ equals to 1. Otherwise, since $B_{n,1} = B_n$, proposition recovers the definition of $L_p(s,\omega^i)$.

where $B_n\left(\frac{a}{f}\right)$ is the value of $n$-th Bernoulli polynomial at $\frac{a}{f}$.

We now turn to the construction of the unit $u$.

Fix a primitive root $g$ modulo $p$. For instance $5$ for $p=37$.
Consider some cyclotomic unit
$$
\xi_g=\zeta_p^{\frac{1-g}{2}}\frac{1-\zeta_p^g}{1-\zeta_p}.
$$
By the choice of the factor $\zeta_p^{\frac{1-g}{2}}$, it is invariant under complex conjugation.

However, we need this unit to satisfy some additional properties. More precisely, we want it to admit a $p$-th root locally at the prime above $p$. This can be achieved by projecting the cyclotomic unit onto the appropriate isotypic component, using the relation between $\omega^i$-components of cyclotomic units and special values of $L_p(s,\omega^i)$.

Let's consider $\operatorname{pr}_{\omega^i,N}$ -- truncation modulo $p^N$ of the $\operatorname{pr}_{\omega^i} = \sum\limits_{a=1}^{p-1}\sigma_a\omega^{-i}(a)$ -- projector onto $\omega^i$-isotypical component.

If we denote by the $\omega^{i}_N(a)$ truncation of the $p$-adic number $\omega^{i}(a)$ modulo $p^N$ then the truncation of the projector can be written in the next form:
$$\operatorname{pr}_{\omega^i,N} = \sum\limits_{a=1}^{p-1}\sigma_a\omega^{-i}_N(a)$$

Finally, the unit $u$ described above is constructed as follows:

$$
u = \operatorname{pr}_{\omega^i,N}(\xi_g),
$$
where $i$ -- irregular index and $N$ is sufficiently large.

We now turn to the justification of this choice of the unit.

The following is the key theorem for it.

**Definition**. The group of cyclotomic units $C$ is the subgroup of units generated by $(1-\zeta_p^a)/(1-\zeta_p)$.

Denote by $\pi$ unique principal prime ideal $(1-\zeta_p)$ lying above $p$.
Let $U_{1,\pi}$ be the group of principal $\pi$-local units of $K$ and $C_{1,\pi} = (C \otimes \mathbb{Z}_p) \cap U_{1,\pi}$.

Note that for sufficciently large $N$ unit $u$ generates corresponding component $C_{1,\pi}^{(i)} = \mathrm{pr}_{\omega^i} C_{1,\pi}$.

**Theorem**.
$$
v_p\left(\left[U_{1,\pi}^{(i)} : C_{1,\pi}^{(i)} \right]\right) = v_p\left(L_p(1,\omega^i)\right)
$$

This theorem can be proven by analyzing the following formula for value of $L_p(s,\omega^i)$ at $s=1$:

**Theorem**.
$$
L_{p}(1,\omega^{i})=-\left(1-\frac{1}{p}\right)\frac{\tau(\omega^{i})}{p}\sum\limits^{p}_{a=1}\omega^{-i}(a)\log_{p}(1-\zeta^a)
$$

Thus it remains to check that the corresponding value at $s=1$ is divisible by $p$. This follows from the following congruence.

**Theorem**. Let $\omega^i$ be a non-trivial power of the Teichmüller character. Then, for any $m,n \in \mathbb{Z}$, one has
$$
L_p(m,\omega^i) \equiv L_p(n,\omega^i) \pmod p.
$$

**Corollary**. Suppose $i$ is even. Then $p \mid B_{i}$ if and only if $v_p(L_p(1, \omega^i)) > 0$.

We know that the value of this $p$-adic $L$-function at the interpolation point corresponding to an irregular index is divisible by $p$, since the corresponding Bernoulli number is divisible by $p$. Hence its value at $s=1$ is also divisible by $p$.

Notice that this argument only uses the divisibility of special values by $p$, and not the existence of a zero of the $p$-adic $L$-function. The latter is a stronger statement, which will be discussed later in connection with the Main Conjecture.

But two difficulties can be eliminated if we assume Kummer-Vandiver conjecture.

We have already seen that the Kummer extension constructed from $u$ is non-trivial. The Kummer--Vandiver conjecture predicts that this is the general situation.

**Kummer--Vandiver conjecture (first form)**. The prime $p$ does not divide the class number of the maximal real subfield $K^+$ of $K=\mathbb{Q}(\zeta_p)$.

By the analytic class number formula, this statement is equivalent to the following one.

**Kummer--Vandiver conjecture (second form)**. The prime $p$ does not divide the index of the group of cyclotomic units in the full unit group modulo roots of unity of $K$.

Cyclotomic units of $K$ genereated by $\frac{1-\zeta_p^a}{1-\zeta}$ for $a=2,\cdots,\frac{p-1}{2}+1$.

The following script verifies the Kummer--Vandiver conjecture for primes small enough that a system of fundamental units of $K$ can be computed.

```gp
z=Mod(x,polcyclo(p));

L = List();

for(aa=2,(p-1)/2+1,if(aa%p!=0,listput(L,(1-z^aa)/(1-z))));

C=Vec(L); M=Mat();

for(ii=1,#C,M=concat(M,[bnfisunit(K,C[ii])[i] | i<-[1..(p-1)/2-1]]~));

IM=matimage(M);
matdet(IM)
```
