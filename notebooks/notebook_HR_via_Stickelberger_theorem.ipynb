## 1. The Herbrand–Ribet theorem

Let $p$ be an odd prime, $F=\mathbb{Q}(\mu_p)$, $\Delta=\operatorname{Gal}(F/\mathbb{Q})\cong(\mathbb{Z}/p\mathbb{Z})^\times$, and let $A$ be the $p$-part of the ideal class group of $F$.

Let $\omega:\Delta\to\mathbb{Z}_p^\times$ be the Teichmüller character. Since $\#\Delta=p-1$ is prime to $p$, $A$ splits into eigenspaces

$$A=\bigoplus_{i \bmod (p-1)} A^{\omega^i},\qquad A^{\chi}=e_\chi A,\qquad e_\chi=\frac{1}{p-1}\sum_{\sigma\in\Delta}\chi^{-1}(\sigma)\,\sigma .$$

**Theorem (Herbrand 1932; Ribet 1976).** For odd $i$ with $3\le i\le p-2$:

$$A^{\omega^i}\neq 0 \iff p \mid B_{p-i} \left( \iff p\mid B_{1,\omega^{-i}}\right),$$

where $B_{1,\psi}=\frac1p\sum_{a=1}^{p-1}a\,\psi(a)$, and the last equivalence is the congruence $B_{1,\omega^{j}}\equiv B_{j+1}/(j+1)\pmod p$ for $j\not\equiv 0,-1 \pmod{p-1}$.

Herbrand's direction ($\Rightarrow$) is a corollary of Stickelberger's theorem. We have already discussed an approach to Ribet’s converse to Herbrand’s theorem using cyclotomic units. Kolyvagin's Euler system of Gauss sums, in Rubin's exposition, reproves the converse and in fact gives the exact order $\#A^{\omega^i}=p^{v_p B_{1,\omega^{-i}}}$.

**Example $p=37$.** The only even $j\le 37$ with $37\mid B_j$ is $j=32$, so $(37,32)$ is the unique irregular pair, $i=p-j=5$ is the unique irregular odd index, and in fact $A=A^{\omega^5}\cong\mathbb{Z}/37\mathbb{Z}$ (the class number of $\mathbb{Q}(\zeta_{37})$ is exactly $37$).

## 2. Stickelberger's theorem

Let $\tau_a:\zeta\mapsto\zeta^a$, and define

$$\theta = \sum_{\substack{a\bmod p\\ (a,p)=1}} \Big\{\frac{a}{p}\Big\}\tau_a^{-1} \in \mathbb{Q}[\Delta].$$

Fix $c$ primitive root modulo $p$ (for $p=37$ choose $c=2$), and define

$$\theta(1)=(\tau_c-c)\theta$$

**Theorem (Stickelberger).** $\theta(1)$ annihilates the ideal class group of $\mathbb{Q}(\zeta_{p})$.

## 3. Herbrand's theorem as a corollary of Stickelberger's theorem

Denote by $\chi$ some $\mathbb{Z}_p^{\times}$-valued character of $\Delta$. (For $p=37$ let $\chi=\omega^5$).

Project $\theta(1)$ to the $\chi$-eigenspace for an odd $\chi=\omega^k\ne\omega$. A direct computation gives $$e_\chi\theta = B_{1,\chi^{-1}}\,e_\chi,$$ so $(\chi(c)-c)\,B_{1,\chi^{-1}}$ annihilates $A^{\chi}$. Choosing $c$ a primitive root makes $c-\chi(c)\equiv c-c^{k}\not\equiv0\pmod p$ (as $k\neq1$), a unit. Hence

$$B_{1,\chi^{-1}}\ \text{annihilates}\ A^{\chi}
\qquad\Longrightarrow\qquad
A^{\omega^i}\neq0 \Rightarrow p\mid B_{1,\omega^{-i}} \equiv \frac{B_{p-i}}{p-i}\ (\mathrm{mod}\ p),$$

which is exactly Herbrand's theorem.

## 4. Further observation

For auxiliary prime $\ell \equiv 1 \pmod{p^4}$ choose $b_\ell\in\mathbb{Z}$ with $\zeta^{\tau_c}=\zeta^{\,b_\ell}$ for all $\zeta\in\mu_{p^4\ell}$, and set

$$\theta(\ell) = (\tau_c-b_\ell) \sum_{\substack{a\bmod p\ell\\ (a,p\ell)=1}} \Big\{\frac{a}{p\ell}\Big\}\tau_a^{-1} \in \mathbb{Z}[\operatorname{Gal}(\mathbb{Q}(\zeta_{p\ell})/\mathbb{Q})].$$

Fix a generator $\sigma_\ell$ of $G_\ell=\operatorname{Gal}\big(\mathbb{Q}(\zeta_{p\ell})/\mathbb{Q}(\zeta_{p})\big)$ and set

$$D_\ell=\sum_{i=1}^{\ell-2} i\,\sigma_\ell^{\,i},\quad N_n=\sum_{\tau\in G_\ell}\tau,  \qquad (\sigma_\ell-1)\,D_\ell=(\ell-1)-N_\ell.$$

Define derived Stickelberger element as an element $\delta(\ell) \in \mathbb{Z}/M\mathbb{Z}$ such that

$$D_\ell e_\chi \theta(\ell) \equiv \delta(\ell) e_\chi N_\ell \pmod{M}$$

We can compute this elements $\delta(\ell)$ using PARI and observe the following: if $\delta(\ell)$ is a unit, then the corresponding prime above $\ell$ is non principal.

$\ell$ -- $\delta(\ell)$ -- $\delta(\ell)$ is unit -- prime above $\ell$ is non principal

## 5. Rubin's theorems

Let $d(\ell)$ be the largest divisor of $M$ dividing $\delta(\ell)$.

**Corollary (Rubin, Cor. 4.2 in [KSGS]).** Let $\ell$ be either auxiliary prime or $1$, let $B\subseteq A$ be the subgroup generated by the classes of primes dividing $\ell$. Then $\#\big(A^\chi/B^\chi\big) \mid d(\ell)$. 

In particular ($\ell=1$): 

* $\#A^\chi \mid d(1)$, and $d(1)\sim B_{1,\chi^{-1}}$. 

* Combined with the analytic class number formula this yields $\#A^\chi=p^{v_p B_{1,\chi^{-1}}}$ for all odd $\chi\ne\omega$ — in particular Ribet's converse to Herbrand.

**Example $p=37$.** $$\#A^{\omega^5}=p^{v_p B_{1,\omega^{31}}}=p^{v_p B_{32}}$$
