```gp
p=37;
```

Classical $L$-functions

```gp
BERN(n,i)=if(i%(p-1)!=0,p^(n-1)*sum(aa=1,p-1,teichmuller(aa+O(p^5))^i*bernpol(n,aa/p)),bernfrac(n)+O(p^5));

LFUN(m,i)=if((i-(lift(1-m+O(p^5))))%(p-1)!=0,-BERN(lift(1-m+O(p^5)),i-(lift(1-m+O(p^5))))/(lift(1-m+O(p^5))),-(1-p^(lift(1-m+O(p^5))-1))*bernfrac(lift(1-m+O(p^5)))/(lift(1-m+O(p^5)))+O(p^5));
```

```gp
for(nn=1,10,print(1-nn, " --- ", LFUN(1-nn,32)))
```

```gp
for(ii=1,p-1,print(ii, " --- ", LFUN(0,ii)))
```

Truncated $L$-functions

```gp
S(k,i,p)=sum(nn=1,p-1,nn^k*teichmuller(nn+O(p^10))^i);

for(ii=1,p-1,print(ii," --- ",S(1,ii,p)+O(p^5)));
```

```gp
for(ii=1,p-1,print(ii, " --- ", LFUN(-1,ii), " --- ",S(1,ii,p)+O(p^5)))
```

```gp
for(ii=1,p-1,print(ii, " --- ", valuation(LFUN(0,ii),p), " --- ",valuation(S(1,ii,p)+O(p^5),p)))
```
